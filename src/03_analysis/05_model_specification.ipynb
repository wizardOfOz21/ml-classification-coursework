### 3.5 Выбор спецификации модели

**Входы**

- Результаты EDA, попарного анализа и проверки гипотез
- Постановка задачи и критерии оценивания

**Выходы**

- Варианты спецификации: состав признаков, обработка, семейства моделей и настройки для сравнения

**Что делаем**

Определим варианты устройства будущей модели на основе накопленных выводов

#### Какие варианты спецификации сравним

| Вариант | Основная идея |
| --- | --- |
| Dummy | Случайно назначает классы с частотами из train |
| Логистическая регрессия | Складывает взвешенные вклады признаков и переводит результат в оценку риска |
| Случайный лес | Усредняет прогнозы множества деревьев решений |
| Градиентный бустинг на деревьях | Добавляет деревья постепенно: каждое уточняет общий результат |

Сравним линейную модель, лес и бустинг с базовым прогнозом Dummy

#### Состав признаков

Фиксируем один набор из 9 признаков для всех моделей. Наборы признаков между собой не сравниваем

| Исключаем | Основание решения |
| --- | --- |
| `loan_percent_income` | Производный от суммы кредита и дохода, которые сохраняем. Часть значений расходится с их отношением сильнее округления |
| `loan_grade` | Сильно связан со ставкой: Spearman ρ = 0,947. Из этой пары сохраняем числовую ставку, чтобы не вводить отдельное кодирование грейда |

Оставляем возраст, доход, тип жилья, стаж, цель и сумму кредита, ставку, прошлые дефолты и длину кредитной истории. Возраст и история связаны, но описывают разные характеристики заёмщика. Пропуски ставки заполним по train\
Это решение об упрощении спецификации; равенство качества с исключёнными признаками не проверяем
