##### Посмотрим на каждый признак отдельно

После очистки осталось 32 411 записей. Начнём с возраста заёмщика — `person_age`

Гистограмма показывает, сколько записей попало в каждый интервал значений. **Медиана** — значение посередине упорядоченного ряда: половина наблюдений не больше неё, половина не меньше

На графике «ящик с усами» (**boxplot**) ящик охватывает среднюю половину наблюдений, линия внутри показывает медиану. Его границы — 25-й и 75-й процентили: ниже них находится 25% и 75% значений. Их разность называют **межквартильным размахом**, или IQR:

$$IQR = Q_3-Q_1.$$

Значения за границами $[Q_1-1{,}5IQR;\;Q_3+1{,}5IQR]$ отмечаем как **статистические выбросы** — необычно далёкие от основной массы. Усы тянутся до крайних наблюдаемых значений внутри этих границ; остальные значения показаны отдельными точками. Это повод посмотреть на них внимательнее, а не автоматически удалить

**Асимметрия** описывает, в какую сторону вытянуто распределение: положительная — вправо, отрицательная — влево

В таблицах Pandas: `count` — число известных значений, `mean` — среднее, `std` — стандартное отклонение, то есть мера разброса вокруг среднего. `min` и `max` — крайние значения, `50%` — медиана
