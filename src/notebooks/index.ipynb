# Модель классификации для задачи оценки риска


## Общая схема и оглавление

Разобьем решение задачи на несколько этапов:

1. **Постановка задачи** — уточняем задачу: смотрим какие нужные признаки, какие у них значения, что предсказываем
2. **Сбор данных** — в нашем случае пропускаем этот этап, так как датасет нам дали как часть формулировки
3. **Анализ, выбор модели и разбиение данных** (объединяем в один этап, так как между указанными шагами не последовательная связь)
   - Базовый EDA: общий обзор, корректность, одномерный анализ
   - Разбиение данных на train / validation / test
   - Корреляционный анализ на train: двумерный + проверка многомерных гипотез
   - Кандидаты моделей и протокол сравнения
4. **Финальная предобработка** — подгтавливаем данные с учетом выбранной модели и выводов EDA
5. **Обучение** — выполняем алгоритм обучения согласно выбранной модели, подбираем гиперпараметры на validation-выборке
6. **Оценка** — проверяем готовую модель на test-выборке
7. **Эксплуатация** — применяем на "реальных" данных
