# Модель классификации для задачи оценки риска


## Общая схема и оглавление

Разобьём решение задачи на несколько этапов Датасет — это наша таблица: одна строка описывает один кредит, а признаки — сведения о заёмщике и условиях займа

1. **Постановка задачи** — уточняем, какие признаки у нас есть, что означают их значения и что будем предсказывать
2. **Сбор данных** — в нашем случае пропускаем: датасет дали как часть задания
3. **Анализ, выбор модели и разбиение данных** — объединяем эти шаги, потому что решения на одном из них могут влиять на другие
   - Базовый **EDA**, то есть разведочный анализ: знакомимся с данными, проверяем ошибки и смотрим на каждый признак отдельно
   - Делим данные на **train** — для обучения, **validation** — для выбора настроек, **test** — для итоговой проверки Каждую такую часть называют выборкой
   - На train смотрим на связи между парами признаков и проверяем предположения о сочетаниях нескольких признаков
   - Выбираем, какие модели попробовать и как их сравнить
4. **Финальная предобработка** — подготавливаем данные с учётом моделей и выводов EDA
5. **Обучение** — обучаем модели и выбираем гиперпараметры по validation Гиперпараметры — настройки алгоритма, которые задаём до обучения
6. **Оценка** — проверяем выбранное решение на test
7. **Эксплуатация** — показываем, как получить прогноз для новой заявки
