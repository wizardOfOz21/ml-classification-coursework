## 2. Сбор данных

**Входы**

- Требования к данным
- Доступные источники

**Выходы**

- Датасет для анализа и обучения модели

**Что делаем**

Обычно на этом этапе собираем данные из доступных источников по сформулированным требованиям и объединяем их в датасет

В нашем задании требования уже указывают на конкретный готовый датасет — Credit Risk Dataset с Kaggle. Его предоставили вместе с заданием, поэтому отдельный сбор данных здесь не требуется
